14. Read the image ('boy.jpg').

15. Adjust the brightness of the image.

16. Create brighter and darker images.

```python
#img_brighter = cv2.add(img, matrix)
#img_darker = cv2.subtract(img, matrix)
# YOUR CODE HERE
matrix_ones = np.ones(boyy.shape, dtype=np.float64)
b_img = (boyy + matrix_ones * 50).astype('uint8')
d_img = (boyy - matrix_ones * 50).astype('uint8')
```

18. Modify the image contrast.

```python
# Create two higher contrast images using the 'scale' option with factors of 1.1 and 1.2 (without overflow fix)
# matrix1 = 
# matrix2 = 
# img_higher1 = 
# img_higher2 = 
# YOUR CODE HERE
matrix1 = np.ones(boyy.shape, dtype=np.float64) * 1.1
matrix2 = np.ones(boyy.shape, dtype=np.float64) * 1.2

img_higher1 = boyy * matrix1
img_higher2 = boyy * matrix2
```

19. Display the images (Original, Lower Contrast, Higher Contrast).

```python
plt.figure(figsize=(9,3))
plt.subplot(1,3,1)
plt.imshow(boyy.astype('uint8'))
plt.subplot(1,3,2)
plt.imshow(img_higher1.astype('uint8'))
plt.subplot(1,3,3)
plt.imshow(img_higher2.astype('uint8'))
```

20. Split the image (boy.jpg) into the B,G,R components & Display the channels.

```python
boyyy=hsv = cv2.cvtColor(boyy, cv2.COLOR_RGB2BGR)
B, G, R = cv2.split(boyyy)

plt.figure(figsize=(9,3))

plt.subplot(1,3,1)
plt.imshow(B)
plt.title("Blue Channel")
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(G)
plt.title("Green Channel")
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(R)
plt.title("Red Channel")
plt.axis('off')

plt.show()
```

21. _merged the R, G, B , display along with orginal image 

```python
B, G, R = cv2.split(boyy)
merged = cv2.merge((B, G, R))
plt.figure(figsize=(6,3))

plt.subplot(1,2,1)
plt.imshow(boyy)
plt.title("Original Image")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(merged)
plt.title("Merged Image")
plt.axis('off')

plt.show()
```

22. Split the image into the H, S, V components &  Display the channels.

```python
hsv = cv2.cvtColor(boyy, cv2.COLOR_RGB2HSV)
H, S, V = cv2.split(hsv)
plt.figure(figsize=(9,3))
plt.subplot(1,3,1)
plt.imshow(H)
plt.title("Hue Channel")
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(S)
plt.title("Saturation Channel")
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(V)
plt.title("Value Channel")
plt.axis('off')

plt.show()
```

23. _merged the H, S, V, display along with orginal image 

```python

merged_hsv = cv2.merge((H, S, V))
merged_bgr = cv2.cvtColor(merged_hsv, cv2.COLOR_HSV2BGR)

plt.figure(figsize=(6,3))

plt.subplot(1,2,1)
plt.imshow(boyy)
plt.title("Original Image")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(cv2.cvtColor(merged_bgr, cv2.COLOR_BGR2RGB))
plt.title("Merged HSV Image")
plt.axis('off')

plt.show()
```
